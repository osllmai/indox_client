## Step 3: Initialize the client

```python
from indox_client import IndoxClient

# Option A: From environment variable
client = IndoxClient.from_env()
```

## Step 4: Check supported formats

```python
# Document formats
doc_formats = client.docs.supported_conversions()
print("Document formats:", doc_formats)

# Image formats
img_formats = client.media.image_formats()
print("Image formats:", img_formats)

# Video formats
vid_formats = client.media.video_formats()
print("Video formats:", vid_formats)
```

## Example: Convert a PDF to DOCX

```python
# Convert from URL
result = client.docs.convert_url(
    file_url="https://example.com/sample.pdf",
    target_formats=["docx"]
)
print("Conversion started:", result)
```

```python
# Wait for completion
conversion_id = result["conversion_id"]
completed = client.docs.wait_for_completion(conversion_id, timeout_seconds=120)
print("Conversion completed:", completed)
```

```python
# Download the result
output_path = client.docs.download(
    conversion_id=conversion_id,
    output_path="./output.docx",
    fmt="docx"
)
print(f"Downloaded to: {output_path}")
```

## Example: Convert a local file

```python
# Convert local PDF to DOCX and TXT
result = client.docs.convert_file(
    file_path="./my_document.pdf",
    target_formats=["docx", "txt"]
)
print("Conversion result:", result)
```

## Example: Convert an image

```python
# Convert PNG to WebP and JPG
result = client.media.convert_image(
    file_url="https://example.com/image.png",
    target_formats=["webp", "jpg"]
)
print("Image conversion:", result)

# Wait and download
completed = client.media.wait_for_conversion(result["conversion_id"])
client.media.download_media(
    conversion_id=result["conversion_id"],
    output_path="./output.webp",
    fmt="webp"
)
```

## Error handling

```python
from indox_client import IndoxHTTPError, IndoxClientError

try:
    result = client.docs.convert_url(
        file_url="https://example.com/file.pdf",
        target_formats=["docx"]
    )
except IndoxHTTPError as e:
    print(f"HTTP Error {e.status_code}: {e}")
    print(f"Request ID: {e.request_id}")
except IndoxClientError as e:
    print(f"Client Error: {e}")
```

## Cleanup

```python
# Close the client when done
client.close()
```
